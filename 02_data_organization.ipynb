# 02 | Data Organization

Input: `cleaned_credit_data.csv`  |  Output: `organized_credit_data.csv`

```python
from pathlib import Path
import pandas as pd

source = Path("cleaned_credit_data.csv")
if not source.exists():
    raise FileNotFoundError("Run 01_data_cleaning.ipynb first.")

data = pd.read_csv(source)
target = "loan_status"
categorical_columns = [
    "person_home_ownership",
    "loan_intent",
    "loan_grade",
    "cb_person_default_on_file",
]
for column in categorical_columns:
    data[column] = data[column].astype("category")
data[target] = data[target].astype("int8")

data["income_band"] = pd.cut(
    data["person_income"],
    bins=[-float("inf"), 25000, 50000, 100000, float("inf")],
    labels=["Up to 25k", "25k-50k", "50k-100k", "Above 100k"],
)
data["loan_to_income_band"] = pd.cut(
    data["loan_percent_income"],
    bins=[-float("inf"), 0.10, 0.20, 0.35, float("inf")],
    labels=["Up to 10%", "10%-20%", "20%-35%", "Above 35%"],
)

data.to_csv("organized_credit_data.csv", index=False)
data_dictionary = pd.DataFrame(
    {
        "column": data.columns,
        "data_type": [str(data[column].dtype) for column in data.columns],
        "missing_values": [int(data[column].isna().sum()) for column in data.columns],
        "distinct_values": [int(data[column].nunique(dropna=True)) for column in data.columns],
    }
)
display(data_dictionary)
print(f"Saved organized_credit_data.csv with {len(data):,} rows.")
```
