# 05 | Power BI Preparation

Inputs: outputs from stages 01-04  |  Output folder: `powerbi_exports/`

```python
from pathlib import Path
import pandas as pd

project_root = next(
    (parent for parent in (Path.cwd(), *Path.cwd().parents) if (parent / "credit_risk_dataset.csv").exists()),
    None,
)
if project_root is None:
    raise FileNotFoundError("Open this notebook from the project folder containing credit_risk_dataset.csv.")

output = project_root / "powerbi_exports"
required_files = [
    "portfolio_segments.csv",
    "woe_iv_detail.csv",
    "scored_test_customers.csv",
    "model_performance.csv",
    "feature_importance.csv",
]
missing = [name for name in required_files if not (output / name).exists()]
if missing:
    raise FileNotFoundError(f"Run stages 03 and 04 first. Missing exports: {missing}")

for path in output.glob("*.csv"):
    if path.name not in required_files:
        path.unlink()

data = pd.read_csv(project_root / "organized_credit_data.csv")
scored_test = pd.read_csv(output / "scored_test_customers.csv")
print(f"Power BI exports are ready in {output.resolve()}")
print(f"Portfolio rows: {len(data):,}; scored test rows: {len(scored_test):,}")
print("Dashboard tables:")
for name in required_files:
    print(f" - {name}")
```
