# 01 | Data Cleaning

Input: `credit_risk_dataset.csv`  |  Output: `cleaned_credit_data.csv`

```python
from pathlib import Path
import pandas as pd

source = Path("credit_risk_dataset.csv")
if not source.exists():
    raise FileNotFoundError(f"Could not find {source.resolve()}")

raw = pd.read_csv(source)
target = "loan_status"
required = {target, "person_income", "loan_amnt"}
missing_columns = required - set(raw.columns)
if missing_columns:
    raise ValueError(f"Missing required columns: {sorted(missing_columns)}")
if raw[target].isna().any() or not set(raw[target].unique()).issubset({0, 1}):
    raise ValueError("loan_status must be complete and coded as 0/1")

quality_checks = pd.DataFrame(
    {
        "check": [
            "Rows before cleaning",
            "Exact duplicate rows",
            "Missing employment length",
            "Missing interest rate",
            "Age above 100",
            "Employment length above age",
        ],
        "count": [
            len(raw),
            int(raw.duplicated().sum()),
            int(raw["person_emp_length"].isna().sum()),
            int(raw["loan_int_rate"].isna().sum()),
            int((raw["person_age"] > 100).sum()),
            int((raw["person_emp_length"] > raw["person_age"]).sum()),
        ],
    }
)
display(quality_checks)

cleaned = raw.drop_duplicates().copy()
cleaning_log = pd.DataFrame(
    {
        "rows_before": [len(raw)],
        "exact_duplicates_removed": [len(raw) - len(cleaned)],
        "rows_after": [len(cleaned)],
        "missing_values_remaining": [int(cleaned.isna().sum().sum())],
    }
)
cleaned.to_csv("cleaned_credit_data.csv", index=False)
display(cleaning_log)
print("Saved cleaned_credit_data.csv")
```
